# Rugged fitness landscape schematic

An illustrative random landscape built from a sum of Gaussian bumps. Purely a
diagram: it is deterministic given `SEED` and uses no simulation output.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

REPO = Path.cwd()
while not (REPO / "pyproject.toml").exists() and REPO != REPO.parent:
    REPO = REPO.parent

from propgen.plotting import set_paper_style

set_paper_style()
FIGDIR = REPO / "figures"
FIGDIR.mkdir(exist_ok=True)

SEED = 33
CMAP = LinearSegmentedColormap.from_list(
    "propgen", ["#003f5c", "#7a5195", "#ef5675", "#ffa600"]
)


def random_fitness_landscape(grid_size=150, xlim=(-4, 4), ylim=(-4, 4),
                             n_peaks=10, seed=SEED):
    """A random landscape-like surface: a sum of Gaussian bumps."""
    rng = np.random.default_rng(seed)
    x = np.linspace(*xlim, grid_size)
    y = np.linspace(*ylim, grid_size)
    X, Y = np.meshgrid(x, y)

    Z = np.zeros_like(X)
    for _ in range(n_peaks):
        x0 = rng.uniform(*xlim)
        y0 = rng.uniform(*ylim)
        amp = rng.uniform(0.5, 1.5) * rng.choice([-1.0, 1.0])
        width = rng.uniform(0.4, 1.2)
        Z += amp * np.exp(-((X - x0) ** 2 + (Y - y0) ** 2) / (2 * width ** 2))
    return X, Y, Z


X, Y, Z = random_fitness_landscape()
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z, cmap=CMAP)
ax.set_axis_off()
plt.tight_layout()
plt.savefig(FIGDIR / "landscape_diagram_3d.pdf")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.imshow(Z, cmap=CMAP, origin="lower", extent=(-4, 4, -4, 4))
plt.axis("off")
plt.tight_layout()
plt.savefig(FIGDIR / "landscape_diagram_2d.pdf")
plt.show()
```

# Persister resuscitation and partitioning

The diffusion limit of ProP Gen theory applied to bacterial persistence. A
dormant persister population resuscitates at an accelerating rate and
partitions into failed, damaged and healthy cells.

This figure is a closed-form solution &mdash; it needs no simulation and no
stored data. See `experiments/persister/model.py` for the derivation and
parameters.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

REPO = Path.cwd()
while not (REPO / "pyproject.toml").exists() and REPO != REPO.parent:
    REPO = REPO.parent

sys.path.insert(0, str(REPO / "experiments" / "persister"))
from model import PARAMETERS, compartments

from propgen.plotting import set_paper_style

set_paper_style()
FIGDIR = REPO / "figures"
FIGDIR.mkdir(exist_ok=True)

PARAMETERS
```

```python
t = np.linspace(0.0, 10.0, 400)
fractions = compartments(t)          # starts as all persisters

COLORS = {"P": "#00416c", "F": "#663d36", "D": "#ad6b06", "H": "#607124"}
LABELS = {"P": "Persisters", "F": "Failed", "D": "Damaged", "H": "Healthy"}

plt.figure(figsize=(6, 5))
for key in ("P", "F", "D", "H"):
    plt.plot(t, fractions[key], label=LABELS[key], lw=3, color=COLORS[key])

plt.xlabel("Time")
plt.ylabel("Fraction of population")
plt.xticks([])
plt.ylim(-0.01, 1.05)
plt.legend(fontsize=14)
plt.tight_layout()
plt.savefig(FIGDIR / "persister_fractions.pdf")
plt.show()
```
